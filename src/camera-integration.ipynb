## Camera Integration with model
---

```python
from tensorflow.keras.models import load_model

model = load_model("../model/vgg16_model.keras")
```

```python
def detect_Face_Mask(frame):
    y_pred = model.predict_classes(frame.shape(1,224,224,3))
    return y_pred
```

```python
import cv2

def capture_show():
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Coudn't Read Frame")
            break

        detect_Face_Mask(frame)
        
        cv2.imshow("Webcame", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Exit")
            break

    cap.release()
    cv2.destroyAllWindows()

capture_show()
```
