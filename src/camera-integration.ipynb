## Camera Integration with model
---

```python
import cv2
from tensorflow.keras.models import load_model
```

```python
model = load_model("../model/vgg16_model.keras")
```

```python
def detect_Face_Mask(frame):
    # y_pred = model.predict_classes(frame.reshape(1,224,224,3))
    y_pred = (model.predict(frame.reshape(1,224,224,3)) > 0.5).astype("int32")
    return y_pred[0][0]
```

> Test Code
---

```python
test = cv2.imread('../assets/khikhi.jpg')
test = cv2.resize(test, (224,224))
detect_Face_Mask(test)
```

```python


def capture_show():
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Coudn't Read Frame")
            break
        img = cv2.resize(frame, (224,224))
        y_pred = detect_Face_Mask(img)
        print(y_pred)
        
        cv2.imshow("Webcame", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Exit")
            break

    cap.release()
    cv2.destroyAllWindows()

capture_show()
```
