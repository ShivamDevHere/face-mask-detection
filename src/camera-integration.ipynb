## Camera Integration with model


>Imported Libraries
---

```python
import cv2
from tensorflow.keras.models import load_model
```

> Loaded Models
---

```python
model = load_model("../model/vgg16_model.keras")
```

> Define function for prediction
---

```python
# def detect_Face_Mask(frame):
#     y_pred = model.predict_classes(frame.reshape(1,224,224,3))
#     # y_pred = (model.predict(frame.reshape(1,224,224,3)) > 0.5).astype("int32")
#     return y_pred[0][0]

def detect_Face_Mask(frame):
    y_pred = model.predict(frame.reshape(1, 224, 224, 3), verbose=0)
    return int(y_pred[0][0] > 0.5)
```

> Camera Integration
---

```python


def capture_show():
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Coudn't Read Frame")
            break
        img = cv2.resize(frame, (224,224))
        y_pred = detect_Face_Mask(img)
        print(y_pred)
        
        cv2.imshow("Webcame", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Exit")
            break

    cap.release()
    cv2.destroyAllWindows()

capture_show()
```

> Drawining Definetion
---

```python
def draw_label(img, text, pos, bg_color):
    text_size = cv2.getTextSize(text, cv2.FONT_HERSHEY_COMPLEX, 1, cv2.FILLED)

    end_x = pos[0] + text_size[0][0] + 2
    end_y = pos[1] + text_size[0][1] - 2

    cv2.rectangle(img, pos, (end_x, end_y), bg_color, cv2.FILLED)
    cv2.putText(img, text, pos, cv2.FONT_HERSHEY_COMPLEX, 1, (0,0,0), 1, cv2.LINE_AA)
    
```
