## Concept

> cv2 loads as **BGR** but matplot lib wants **RGB**
---

```python
import matplotlib.pyplot as plt
import cv2

img_bgr = cv2.imread("..\\assets\khikhi.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

ax1.imshow(img_bgr)
ax1.set_title("BGR")
ax1.axis("off")

ax2.imshow(img_rgb)
ax2.set_title("RGB")
ax2.axis("off")
```

> Model Conversion from Functional to Sequential
---

- **Reason:** To remove the original classification layer and customize VGG16 for our specific classification task.

```python
from keras.applications.vgg16 import VGG16
from keras import Sequential

vgg = VGG16()
model = Sequential()

for layer in vgg.layers[:-1]:
    model.add(layer)

model.summary()
```
